bài 1.1

Với x = 23 và y = 6, hãy tính tay giá trị các biểu thức x / y, x // y, x % y, x ** 2, -x // y, -x % y, 2 ** 3 ** 2, (2 ** 3) ** 2; sau đó kiểm tra lại bằng máy.

```python
x = 23
y = 6
print ("x / y = ", x / y)
print ("x // y = ", x//y)
print ("x % y = ", x % y)
print ("x ** y = ", x ** y)
print ("-x // y = ", -x // y)
print ("-x % y = ", -x % y)
print ("2 ** 3 ** 2 = ", 2 ** 3 ** 2)
print ("(2 ** 3) ** 2 = ", (2 ** 3) ** 2)
```

bài 1.3

Dự đoán đầu ra của bốn dòng dưới đây rồi chạy để kiểm tra. Chú ý các giá trị làm tròn và phép ép kiểu; giải thích vì sao round(2.5) và round(3.5) không cho kết quả “tăng một đơn vị” như nhau.

Vì round() làm tròn về số chẵn gần nhất

```python
print(round(2.5), round(3.5), round(0.5), round(1.2345, 2))
print(int(-3.7), -3.7 // 1, int(3.99))
print(int("12") + int("3"), "12" + "3")
print("ab" * 3 + "c", 7 / 2 * 2, 10 - 2 ** 2 * 3)
```

1.4

 Dự đoán chính xác những gì chương trình dưới đây in ra (chú ý khoảng trắng, dấu xuống dòng và định dạng số), rồi chạy kiểm tra.

```python
a, b = 7, 2
print(a, b, sep="")
print(a / b, a // b, sep=" | ")
print(f"{a}+{b}={a + b}", end="; ")
print(f"{a / b:.1f}")
print("a" "b", 3 * "ab")
print(f"{1234.5678:,.2f}", f"{5:03d}", f"{0.256:.1%}")
```

bài 1.5

Đoạn mã dưới đây tính chỉ số cơ thể BMI = cân nặng (kg) / chiều cao (m)^2. chạy với đầu vào 60 và 1.7, đọc thông báo giải thích ít nhất 2 lỗi khác nhau rồi sửa để BMI = 20,8

giải thích:

- Hàm input() trả về giá trị dạng chuỗi, khi dùng với dạng số ta dùng thêm int(input), float(input),...
- phép tính luỹ thừa bị sai, phải dùng " ** ".

```python
can_nang = input("Cân nặng (kg): ")
chieu_cao = input("Chiều cao (m): ") 
bmi = can_nang / chieu_cao ^ 2       
print(f"BMI = {bmi:.1f}")
```

```python
can_nang = float(input("Cân nặng (kg): "))
chieu_cao = float(input("Chiều cao (m): "))
bmi = can_nang / (chieu_cao) ** 2
print(f"BMI = {bmi:.1f}")
```
