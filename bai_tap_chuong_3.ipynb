## TH 3.2

Điểm của $8$ giám khảo là `[7.5, 9.0, 3.0, 8.0, 6.5, 10.0, 7.0, 2.5]`. Để giảm ảnh hưởng của giá trị bất thường, bỏ đi $n=2$ điểm cao nhất và $n=2$ điểm thấp nhất rồi tính trung bình các điểm còn lại (*trung bình cắt đầu--cuối*). So sánh với trung bình thông thường; *không* sửa danh sách gốc.

```python
ds = [7.5, 9.0, 3.0, 8.0, 6.5, 10.0, 7.0, 2.5]

tb = sum(ds)/len(ds)
sap_xep = sorted(ds)
cat = sap_xep[2:-2]
tbm = sum(cat)/len(cat)

print("danh sách điểm: ", ds)
print("danh sách điểm đã cắt: ", cat)
print("trung bình danh sách điểm: ", tb)
print("trung bình danh sách điểm đã cắt", tbm)
```

```python
import statistics
ds = [(1, 4.0), (2, 5.5), (3, 6.0), (4, 7.5), (5, 8.0)]

x = []
y = []
for item in ds:
    x.append(item[0])
    y.append(item[1])
    
x_tb = sum(x) / len(x)
y_tb = sum(y) / len(y)
tu_so = sum((xi - x_tb) * (yi - y_tb) for xi, yi in zip(x, y))
mau_so = sum((xi - x_tb) ** 2 for xi in x)
b = tu_so / mau_so
a = y_tb - b * x_tb

du_doan = a + b * 6

print(f"đường hồi quy: y = {a:.2f} + {b:.2f}x")
print(f"dự đoán điểm khi ôn 6 giờ: {du_doan:.2f}")

lr = statistics.linear_regression(x, y)
print(f"hàm statistics.linear_regression: y = {intercept} + {slope} * x")
```

## TH 3.6

Cho danh sách các cặp `(tên, điểm)`: `[("An", 9.0), ("Binh", 7.5), ("Chi", 8.6), ("Dung", 5.0), ("Em", 6.0), ("Phuc", 3.5), ("Giang", 7.0)]`. Quy đổi sang điểm chữ ($\ge8{,}5$: A; $\ge7$: B; $\ge5{,}5$: C; $\ge4$: D; còn lại F) rồi gom tên sinh viên theo điểm chữ bằng từ điển (dùng `defaultdict(list)`). In các nhóm theo thứ tự chữ cái.

```python
from collections import defaultdict
ds = [("An", 9.0), ("Binh", 7.5), ("Chi", 8.6), ("Dung", 5.0), ("Em", 6.0), ("Phuc", 3.5), ("Giang", 7.0)]
def quy_doi_diem(d):
    if d >= 8.5:
        return "A"
    if d >= 7:
        return "B"
    if d >= 5.5:
        return "C"
    if d >= 4:
        return "D"
    return "F"
nhom = defaultdict(list)
for ten, diem in ds:
    nhom[quy_doi_diem(diem)].append(ten)

for chu in sorted(nhom):
    print(f"{chu}: {nhom[chu]}")
```

## TH 3.8

Doanh thu theo sản phẩm của hai cửa hàng: \texttt{{"ao": 120, "quan": 80, "non": 30}} và \texttt{{"ao": 90, "giay": 150, "non": 20}}. Hợp nhất thành tổng doanh thu theo sản phẩm, in theo thứ tự giảm dần; tìm các sản phẩm bán ở cả hai cửa hàng và các sản phẩm chỉ bán ở một cửa hàng (dùng phép toán tập hợp trên khoá).

```python
from collections import Counter

ch1 = {"ao": 120, "quan": 80, "non": 30}
ch2 = {"ao": 90, "giay": 150, "non": 20}
tong = Counter(ch1) + Counter(ch2)

print("tổng doanh thu theo sản phẩm giảm dần:")
for sp, dt in sorted(tong.items(), key=lambda kv: kv[1], reverse=True):
    print(f"  {sp}: {dt}")

ca_hai = ch1.keys() & ch2.keys()      
mot_cua_hang = ch1.keys() ^ ch2.keys()
print("bán ở cả hai cửa hàng:", sorted(ca_hai))
print("chỉ bán ở một cửa hàng:", sorted(mot_cua_hang))
```

## TH 3.10 ∗

Dùng sàng Eratosthenes (cài đặt bằng `set`, theo Mệnh đề \ref{md:sang}) tìm các số nguyên tố không vượt quá $1000$: in số lượng, tổng, và số cặp *số nguyên tố sinh đôi* $(p,p+2)$ cùng $\le1000$. Kiểm tra: có $168$ số nguyên tố, tổng bằng $76 127$.

```python
N = 1000
nguyen_to = set(range(2, N + 1))

for p in range(2, int(N ** 0.5) + 1):
    if p in nguyen_to:
        nguyen_to -= set(range(p * p, N + 1, p))

print("ố lượng số nguyên tố:", len(nguyen_to))
print("Tổng:", sum(nguyen_to))

sinh_doi = sorted((p, p + 2) for p in nguyen_to if p + 2 in nguyen_to)
print("Số cặp sinh đôi:", len(sinh_doi))
print("Một vài cặp đầu:", sinh_doi[:5])

assert len(nguyen_to) == 168
assert sum(nguyen_to) == 76127
print("Kiểm tra đạt: 168 số nguyên tố, tổng 76127")
```
