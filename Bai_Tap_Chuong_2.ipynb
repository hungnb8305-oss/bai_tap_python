## TH 2.3

Viết chương trình tính lương tuần như Ví dụ \ref{vd:luong} (lương $100 000$ đồng/giờ, giờ vượt $40$ tính gấp $1{,}5$) nhưng *nhập lại* cho đến khi người dùng gõ một số *không âm*: chữ hoặc số âm đều phải báo lỗi và hỏi lại. Kiểm tra với dãy gõ `abc`, `-2`, `45`.

```python
while True:
    try:
        gio_lam = float(input("nhap so gio lam: "))
        luong_gio = float(input("nhap muc luong: "))
    except ValueError:
        print("Lỗi: bạn phải nhập dữ liệu số.")
        continue
    if gio_lam < 0 or luong_gio < 0:
        print("Lỗi: số giờ và mức lương không được âm.")
        continue
    break
    
if gio_lam <= 40:
    luong = gio_lam * luong_gio
else:
    luong = 40 * luong_gio + (gio_lam - 40) * luong_gio * 1.5

print(f"Tiền lương: {luong:,.0f} đồng")
```

## TH 2.4

Nhập liên tục các số thực (bỏ qua và báo giá trị không phải số) cho đến khi gõ `xong`. In số giá trị hợp lệ, tổng, trung bình, giá trị lớn nhất và nhỏ nhất; nếu chưa nhập số nào thì báo ``chưa có dữ liệu''. Kiểm tra với dãy gõ: `7.5, 8, abc, 6.5, 9, xong`.

```python
so_luong = 0
tong = 0
lon_nhat = None
nho_nhat = None

while True:
    s = input("Nhập một số (gõ 'xong' để kết thúc): ")
    if s == "xong":
        break
    try:
        x = float(s)
    except ValueError:
        print("Bỏ qua: '" + s + "' không phải là số.")
        continue
    so_luong += 1
    tong += x
    if lon_nhat is None or x > lon_nhat:
        lon_nhat = x
    if nho_nhat is None or x < nho_nhat:
        nho_nhat = x
if so_luong == 0:
    print("Chưa có dữ liệu")
else:
    print(f"Số giá trị hợp lệ: {so_luong}")
    print(f"Tổng: {tong}")
    print(f"Trung bình: {tong / so_luong}")
    print(f"Lớn nhất: {lon_nhat}")
    print(f"Nhỏ nhất: {nho_nhat}")
```
